```python
menu={'Paneer Tikka':180, 'Veg Burger':90, 'Margherita Pizza':250, 'White Sauce Pasta':150, 'Dal Makhani':170, 'Butter Naan':40, 'Cold Coffee':80, 'Gulab Jamun':60}
items=list(menu)
order={}
```

```python
while True:
    print()
    print('''WELCOME TO THE RESTAURANT
1) Display Menu
2) Order Food
3) Generate Order Summary
4) Quit''')
    n=input('Enter your choice: ')
    if not n.isdigit():
        print('Enter a number from 1 to 4')
        continue
    n=int(n)
    if n==1:
        print()
        for i in range(len(items)):
            print(i+1, items[i], '- Rs', menu[items[i]])
        print()
        print('Rs 500 or more = 10% off')
        print('Rs 1000 or more = 20% off')
    elif n==2:
        while True:
            print()
            for i in range(len(items)):
                print(i+1, items[i], '- Rs', menu[items[i]])
            p=input('Item number (0 to stop): ')
            if not p.isdigit():
                print('Wrong input')
                continue
            p=int(p)
            if p==0:
                break
            if p>len(items):
                print('No such item')
                continue
            q=input('Quantity of '+items[p-1]+': ')
            if not q.isdigit() or int(q)==0:
                print('Quantity should be 1 or more')
                continue
            item=items[p-1]
            if item in order:
                order[item]+=int(q)
            else:
                order[item]=int(q)
            print(item, 'added')
    elif n==3:
        if len(order)==0:
            print('Nothing ordered yet')
        else:
            print()
            print('      ORDER SUMMARY      ')
            total=0
            count=0
            for item in order:
                print(item, ':', order[item], 'x', menu[item], '=', order[item]*menu[item])
                total+=order[item]*menu[item]
                count+=order[item]
            if total>=1000:
                disc=20
            elif total>=500:
                disc=10
            else:
                disc=0
            cut=total*disc/100
            pay=total-cut

            print('                         ')
            print('Total items =', count)
            print('Bill = Rs', total)
            print('Discount', disc, '% = Rs', round(cut, 2))
            print('To pay = Rs', round(pay, 2))
            conf=input('Confirm order? (y/n): ')
            if conf=='y' or conf=='Y':
                print('Order placed, thank you!')
                order={}
            else:
                print('Order not placed, it is still saved')

    elif n==4:
        print('Thank you, visit again!')
        break

    else:
        print('Enter a number from 1 to 4')
```
